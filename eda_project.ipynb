```python
#Univariate
status_count=df["Status"].value_counts()
status_count
```

```python
#Bivariate
age_consider=(
    df.groupby("Age")["Status"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)
age_consider
```

```python
category_dashboard = (
    df.groupby("Marital Status")["Age"]
      .sum()
      .reset_index()
      .sort_values("Age", ascending=False)
      .head(10)
)
```

```python
fig = px.line(
    category_dashboard,
    x="Marital Status",
    y="Age",
    markers=True,
    title="Monthly Survey Trend"
)

fig.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

numeric_df.corr()
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], bins=20, kde=True)

plt.title("Patient Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.show()
```

```python
print("Average Age:", df["Age"].mean())

print("Average Tumor Size:", df["Tumor Size"].mean())

print("\nPatients by Status:")
print(df["Status"].value_counts())

print("\nAverage Tumor Size by Status:")
print(df.groupby("Status")["Tumor Size"].mean())

print("\nAverage Survival Months by Status:")
print(df.groupby("Status")["Survival Months"].mean())

print("\nPatients by Cancer Stage:")
print(df["6th Stage"].value_counts())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12,5))

sns.countplot(data=df, x="Estrogen Status", ax=axes[0])
axes[0].set_title("Estrogen Status")

sns.countplot(data=df, x="Progesterone Status", ax=axes[1])
axes[1].set_title("Progesterone Status")

plt.tight_layout()
plt.show()
```
